```python
import os

csv_files = [f for f in os.listdir(".") if f.lower().endswith(".csv")]
csv_files
```

```python
import re
def extract_se_meta(csv_files, max_lines = 60):
    meta = {}
    with open(csv_files, "r", encoding = "utf-8", errors = "replace") as f:
      for _ in range(max_lines):
        line = f.readline()
        if not line:
          break
        m = re.match(r'"\s*([^"]+)\s*"\s*,\s*"\s*([^"]+)\s*"\s*$', line.strip())
        if m:
          meta[m.group(1)] = m.group(2)
    return meta
```

```python
import pandas as pd

records = []

for f in csv_files:
  if "__" not in f:
    continue

  concept, variant = f.replace(".csv", "").split("__", 1)

  meta = extract_se_meta(f)

  if "concordance size" not in meta:
    print(f"!! skipped (no concordance size available): {f}")
    continue

  records.append({
    "concept" : concept,
    "variant" : variant,
    "hits" : int(meta["concordance size"]),
    "export_file": f
  })

  hits_df = pd.DataFrame(records)
  hits_df
```

```python
hits_df.to_csv("variant_hits_example.csv", index = False)
print("wrote variant_hits.csv")
```
