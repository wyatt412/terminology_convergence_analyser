```python
import pandas as pd

sample = pd.read_csv("wasm_sample_labeled.csv", skiprows = 4)
sample.head()
```

```python
sample["is_hashtag"].value_counts()
```

```python
import math

n = len(sample)
k = int(sample["is_hashtag"].sum())
p = k / n

p
```

```python
se = math.sqrt(p * (1 - p) / n) if n > 0 else 0
p_hi = max(0, p - 1.96 * se)
p_lo = min(1, p + 1,96 * se)

p, (p_hi, p_lo)
```

```python
hits_df = pd.read_csv("variant_hits_example.csv")
hits_df
```

```python
mask = (hits_df["concept"] == "hashtag") & (hits_df["variant"] == "وسم")

N = hits_df.loc[mask, "hits"].iloc[0]

hits_df.loc[mask, "purity_p"] = p
hits_df.loc[mask, "sample_n"] = n
hits_df.loc[mask, "adjusted_hits"] = N * p

hits_df
```

```python
hits_df.to_csv("variant_hits_example.csv", index = False)
print("variant_hits.csv updated with purity correction")
```
